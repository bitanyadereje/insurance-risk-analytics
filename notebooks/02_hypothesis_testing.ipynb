```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind, chi2_contingency
import warnings
warnings.filterwarnings('ignore')

# Load data
df = pd.read_csv('../data/insurance_data.csv', sep='|', encoding='latin1')
df['ClaimOccurred'] = (df['TotalClaims'] > 0).astype(int)
df['Margin'] = df['TotalPremium'] - df['TotalClaims']

prov_g = df[df['Province'] == 'Gauteng']['TotalClaims']
prov_g = prov_g[prov_g > 0]
prov_wc = df[df['Province'] == 'Western Cape']['TotalClaims']
prov_wc = prov_wc[prov_wc > 0]
t_stat, p_prov = ttest_ind(prov_g, prov_wc, equal_var=False)
print(f"Province (Gauteng vs Western Cape) claim severity p-value: {p_prov:.6f}")
print("Reject H0: Yes" if p_prov < 0.05 else "Reject H0: No")

zips = df['PostalCode'].value_counts().head(2).index
zip1 = df[df['PostalCode'] == zips[0]]['ClaimOccurred']
zip2 = df[df['PostalCode'] == zips[1]]['ClaimOccurred']
cont = pd.crosstab(df['PostalCode'], df['ClaimOccurred'])
chi2, p_zip, dof, ex = chi2_contingency(cont)
print(f"Zip code claim frequency p-value: {p_zip:.6f}")

# 3. Margin difference between same zip codes
margin1 = df[df['PostalCode'] == zips[0]]['Margin']
margin2 = df[df['PostalCode'] == zips[1]]['Margin']
t_stat, p_margin = ttest_ind(margin1, margin2, equal_var=False)
print(f"Margin difference p-value: {p_margin:.6f}")

# 4. Gender risk difference
gender_table = pd.crosstab(df['Gender'], df['ClaimOccurred'])
chi2, p_gender, dof, ex = chi2_contingency(gender_table)
print(f"Gender claim frequency p-value: {p_gender:.6f}")
```
